<details><summary>👀 Need the code? Click to reveal (try on your own first!)</summary>

```python
df["bmi"] = (df["weight_kg"] / df["height_m"] ** 2).round(1)

df["visit_date"] = pd.to_datetime(df["visit_date"])
df["month"] = df["visit_date"].dt.month
df["weekday"] = df["visit_date"].dt.day_name()

def triage(temp):
    if pd.isna(temp):
        return "Not recorded"
    elif temp >= 39:
        return "URGENT"
    elif temp >= 37.5:
        return "Fever"
    else:
        return "Normal"

df["triage"] = df["temp_c"].apply(triage)
print(df["triage"].value_counts())
df[["visit_id", "temp_c", "bmi", "weekday", "triage"]].head(3)
```

## Step 9 · One summary table, then merge  *(~6 min)*
**Concept:** .agg, .reset_index, .merge

📖 The district office wants one table: visits, average wait and fees for each ward, plus who runs each ward. That last piece lives in another file.

🎯 **Your task**
1. Build the summary with named calculations.
2. Load **wards.csv** and look at the column both tables share.
3. Merge the two tables on `ward`.

<details><summary>👀 Need the code? Click to reveal (try on your own first!)</summary>

```python
summary = df.groupby("ward").agg(
    visits=("visit_id", "count"),
    avg_wait=("wait_mins", "mean"),
    total_fees=("fee", "sum"),
).round(1).reset_index()

wards = pd.read_csv("wards.csv")
summary = summary.merge(wards, on="ward", how="left")
summary
```
